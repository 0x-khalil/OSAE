# OSAE study on Kaggle (2x T4)

Runs the study from https://github.com/0x-khalil/OSAE (training + final test evaluation + analysis) and publishes the results to https://huggingface.co/0x-Khalil/OSAE.

**Setup (notebook sidebar), once:**
1. *Session options* -> **Accelerator: GPU T4 x2**, **Internet: on** (needs a phone-verified Kaggle account).
2. *Add-ons* -> *Secrets* -> add a secret labelled **`HF_TOKEN`** (Hugging Face token with *write* access) and tick it for this notebook.

**Run:** *Save Version* -> **Save & Run All (Commit)**. It runs in the background (up to 12 h; you can close the browser). Watch the log in the version page.
* `CONFIG = configs/kaggle_fast.yaml`: halved training budget, one session (~6-7.5 h).
* `CONFIG = configs/kaggle.yaml`: full budget, ~2 sessions. If the log ends with **SESSION PAUSED**, run *Save & Run All* again; it resumes from the private repo `0x-Khalil/OSAE-state`.
* **STUDY COMPLETE** at the end of the log = results are published on the Hub.

```python
import os, subprocess, sys

CONFIG = "configs/kaggle_fast.yaml"   # or "configs/kaggle.yaml" for the full training budget (~2 sessions)
REPO = "/kaggle/working/OSAE"

if not os.path.isdir(os.path.join(REPO, ".git")):
    r = subprocess.run(["git", "clone", "--depth", "1", "https://github.com/0x-khalil/OSAE.git", REPO],
                       capture_output=True, text=True)
    print(r.stdout, r.stderr)
    if r.returncode != 0:
        raise SystemExit("git clone failed. In the notebook sidebar: Session options -> Internet -> On "
                         "(requires a phone-verified Kaggle account), then run again.")
else:
    subprocess.run(["git", "-C", REPO, "pull", "--ff-only"])

os.chdir(REPO)
rc = subprocess.call([sys.executable, "-u", "scripts/kaggle_session.py", "--hf_repo", "0x-Khalil/OSAE", "--config", CONFIG])
if rc != 0:
    raise SystemExit(f"kaggle_session.py exited with {rc}; see the log above")
```
